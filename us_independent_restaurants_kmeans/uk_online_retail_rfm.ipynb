```python
df2 = df1.loc[df1['Country'] == 'United Kingdom']
df2
```

```python
df3 = df2[pd.notnull(df2['CustomerID'])]
df3
```

```python
df3.isnull().sum(axis=0)
```

```python
df4 = df3[(df3['Quantity']>0) & (df3['UnitPrice']>0)] 
df4
```

```python
df4.shape
```

```python
for i in df4.columns:
        count = df4[i].nunique()
        print(i, ": ", count)
```

```python
df4['TotalPrice'] = df4['Quantity'] * df4['UnitPrice']

df4
```

```python
df4['InvoiceDate'].max()
```

```python
import datetime as dt

NOW = dt.datetime(2011,12,10)

df4['InvoiceDate'] = pd.to_datetime(df4['InvoiceDate'])

df4
```

```python
rfmtable = df4.groupby('CustomerID').agg({'InvoiceDate': lambda x: (NOW - x.max()).days,
                                          'InvoiceNo': lambda x: len(x),
                                          'TotalPrice': lambda x: x.sum()})

rfmtable
```

```python
rfmtable.rename(columns={'InvoiceDate': 'recency',
                         'InvoiceNo': 'frequency',
                         'TotalPrice': 'monetary'}, inplace=True)
rfmtable
```

```python
quantiles = rfmtable.quantile(q=[0.25,0.5,0.75])

quantiles
```

```python
quantiles = quantiles.to_dict()
quantiles
```

```python
def RScore(x,p,d):
    if x <= d[p][0.25]:
        return 1
    elif x <= d[p][0.50]:
        return 2
    elif x <= d[p][0.75]:
        return 3
    else:
        return 4

def FMScore(x,p,d):
    if x <= d[p][0.25]:
        return 4
    elif x <= d[p][0.50]:
        return 3
    elif x <= d[p][0.75]:
        return 2
    else:
        return 1
```

```python
segmented_rfm = rfmtable

segmented_rfm['r_quartile'] = segmented_rfm['recency'].apply(RScore, args=('recency',quantiles,))

segmented_rfm
```

```python
segmented_rfm['f_quartile'] = segmented_rfm['frequency'].apply(FMScore, args=('frequency',quantiles,))

segmented_rfm
```

```python
segmented_rfm['m_quartile'] = segmented_rfm['monetary'].apply(FMScore, args=('monetary',quantiles,))

segmented_rfm 
```

```python
segmented_rfm['RFMScore']= segmented_rfm.r_quartile.map(str) + segmented_rfm.f_quartile.map(str) + segmented_rfm.m_quartile.map(str)

segmented_rfm
```

```python
segmented_rfm[segmented_rfm['RFMScore'] == '111'].sort_values('monetary', ascending=False)
```

```python
import numpy as np
import matplotlib.pyplot as plt
from mpl_toolkits.mplot3d import Axes3D

x = segmented_rfm['r_quartile']

y = segmented_rfm['f_quartile']

z = segmented_rfm['m_quartile']

fig = plt.figure(figsize=(10, 10))


ax = plt.subplot(projection='3d')


ax.scatter3D(x, y, z, c=(x + y + z))

plt.title("RFM 3D Plot")

ax.set_zlabel('Recency')
ax.set_ylabel('Frequency')
ax.set_xlabel('Monetary')

plt.show()
```
